```python
# Calculate percentage of missing values
missing_percentage = (df.isnull().sum() / len(df)) * 100

# Create a DataFrame
missing_report = pd.DataFrame({
    "Missing Values": df.isnull().sum(),
    "Percentage": missing_percentage
})

# Display only columns having missing values
missing_report[missing_report["Missing Values"] > 0]
```

```python
# Remove columns where all values are missing
df = df.dropna(axis=1, how="all")
```

```python
# Select numerical columns
numeric_cols = df.select_dtypes(include=["int64", "float64"]).columns

# Fill missing values with median
for col in numeric_cols:
    df[col] = df[col].fillna(df[col].median())
```

```python
# Select categorical columns
categorical_cols = df.select_dtypes(include="object").columns

# Fill missing values with mode
for col in categorical_cols:
    df[col] = df[col].fillna(df[col].mode()[0])
```

Reason:

Mode was used because it replaces missing values with the most frequent category.

```python
# Verify missing values after cleaning
print(df.isnull().sum())
```

```python
# Count duplicates before removal
duplicates_before = df.duplicated().sum()

print("Duplicate Rows Before:", duplicates_before)

# Remove duplicate rows
df = df.drop_duplicates()

# Count duplicates after removal
duplicates_after = df.duplicated().sum()

print("Duplicate Rows After:", duplicates_after)
```

Observation:

Duplicate records were removed to improve data quality and avoid repeated information.

```python
# Remove leading and trailing spaces from text columns
for col in categorical_cols:
    df[col] = df[col].astype(str).str.strip()
```

```python
# Convert text into Title Case
for col in categorical_cols:
    df[col] = df[col].str.title()
```

Observation:

Text values were standardized by removing extra spaces and converting them to title case.

```python
# Check unique values
df["Value"].head()
```

```python
# Function to convert Value into numbers
def convert_value(value):

    if pd.isna(value):
        return np.nan

    value = str(value).replace("€", "")

    if "M" in value:
        return float(value.replace("M", "")) * 1000000

    elif "K" in value:
        return float(value.replace("K", "")) * 1000

    else:
        return float(value)

# Apply function
df["Value"] = df["Value"].apply(convert_value)

# Check result
df["Value"].head()
```

Observation:

The Value column contained currency symbols (€) and abbreviations (M, K). These were converted into numeric values for analysis.

```python
# Function to clean Wage column
def convert_wage(value):

    if pd.isna(value):
        return np.nan

    value = str(value).replace("€", "")

    if "M" in value:
        return float(value.replace("M", "")) * 1000000

    elif "K" in value:
        return float(value.replace("K", "")) * 1000

    else:
        return float(value)

# Apply function
df["Wage"] = df["Wage"].apply(convert_wage)

# Check result
df["Wage"].head()
```

```python
# Apply same function
df["Release Clause"] = df["Release Clause"].apply(convert_value)

# Display first rows
df["Release Clause"].head()
```

```python
# Remove cm symbol and convert to integer
df["Height"] = df["Height"].str.replace("cm", "", regex=False)

df["Height"] = pd.to_numeric(df["Height"])
```

```python
# Remove kg symbol
df["Weight"] = df["Weight"].str.replace("kg", "", regex=False)

# Convert into numeric
df["Weight"] = pd.to_numeric(df["Weight"])
```

```python
# Convert Joined into datetime
df["Joined"] = pd.to_datetime(df["Joined"], errors="coerce")

# Check datatype
df["Joined"].dtype
```

```python
# Function to convert Hits
def clean_hits(value):

    if pd.isna(value):
        return np.nan

    value = str(value)

    if "K" in value:
        return float(value.replace("K", "")) * 1000

    else:
        return float(value)

# Apply function
df["Hits"] = df["Hits"].apply(clean_hits)
```

```python
# Display updated data types
print(df.dtypes)
```

Observation:

The Value, Wage, Release Clause, Height, Weight, Joined, and Hits columns were successfully converted into appropriate numeric and datetime formats. This makes the dataset suitable for further analysis and machine learning.

```python
# Select numeric columns
numeric_columns = df.select_dtypes(include=["int64", "float64"]).columns

# Check number of outliers in each numeric column
for col in numeric_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - (1.5 * IQR)
    upper = Q3 + (1.5 * IQR)

    outliers = df[(df[col] < lower) | (df[col] > upper)]

    print(f"{col}: {len(outliers)} outliers")
```

## Outlier Detection

Outliers were detected using the IQR (Interquartile Range) method. Instead of removing all outliers automatically, they were first identified and evaluated because some extreme values may represent real football players with exceptional performance or market value.

```python
# Cap outliers using IQR limits

for col in numeric_columns:

    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)

    IQR = Q3 - Q1

    lower = Q1 - (1.5 * IQR)
    upper = Q3 + (1.5 * IQR)

    df[col] = np.where(df[col] < lower, lower, df[col])
    df[col] = np.where(df[col] > upper, upper, df[col])
```

## Outlier Treatment

Instead of removing outliers, extreme values were capped using the IQR boundaries. This preserves all player records while reducing the influence of extreme values.
